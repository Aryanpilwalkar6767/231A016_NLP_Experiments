```python
print("Name: Aryan Pilwalkar UIN:231A016")
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import string
```

```python
print("Name: Aryan Pilwalkar UIN:231A016")
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
print("Name: Aryan Pilwalkar UIN:231A016")
print(string.punctuation)
```

```python
print("Name: Aryan Pilwalkar UIN:231A016")
#step 1 => preprocesssing
text_lower = text.lower()
print(text_lower)
```

```python
print("Name: Aryan Pilwalkar UIN:231A016")
tokens = word_tokenize(text_lower)
print(tokens)
```

```python
print("Name: Aryan Pilwalkar UIN:231A016")
text_clean = []
for word in tokens:
  if word not in string.punctuation:
    text_clean.append(word)
print(text_clean)
```

```python
print("Name: Aryan Pilwalkar UIN:231A016")
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(tok,pos='v') for tok in text_clean]
lemmas
```

```python
print("Name: Aryan Pilwalkar UIN:231A016")
lemmatizer.lemmatize("walking",pos='n')
```

```python
print("Name: Aryan Pilwalkar UIN:231A016")
morphological_info = {}
for token, lemma in zip(text_clean,lemmas):
  analysis = ""
  if token.endswith("ation"):
    analysis = "Derivational: '{}' + -ation -> result of action".format(token.replace('ation', '')) # Infer base by removing suffix
  elif token.endswith("er") and token != lemma:
    analysis = "Derivational: '{}' + -er -> person who does the action".format(token.replace('er', '')) # Infer base by removing suffix
  elif token.endswith("ing") and token != lemma:
    analysis = "Inflectional: Present participle/continuous form of '{}'".format(lemma)
  elif token != lemma:
    analysis = "Inflectional or irregular form of '{}'".format(lemma)
  else:
    analysis = "Base/root form"

  morphological_info[token] = analysis
```

```python
print("Name: Aryan Pilwalkar UIN:231A016")
morphological_info
```
